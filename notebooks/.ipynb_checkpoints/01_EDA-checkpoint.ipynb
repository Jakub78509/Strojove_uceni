# 01 – Porozumění datům (EDA)

Fáze **Data Understanding** v CRISP-DM (přednáška 01). Postupujeme podle přednášky 03:
typy atributů → statistiky polohy, rozptylu a tvaru → histogramy a boxploty → scatter ploty a heatmapa → multikolinearita → chybějící hodnoty → odlehlé hodnoty.

**Rozhodovací problém:** Skončí jezdec na pódiu (1.–3. místo), když známe informace dostupné **před startem závodu**?
**Instance** = jeden jezdec v jednom závodě. **Cílová třída** = `podium` (ano/ne).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", 50)
DATA = "../data"

races = pd.read_csv(f"{DATA}/races.csv")
results = pd.read_csv(f"{DATA}/results.csv")
qualifying = pd.read_csv(f"{DATA}/qualifying.csv")
print(races.shape, results.shape, qualifying.shape)
```

## 1. Data wrangling – spojení zdrojů
Integrace více zdrojů (přednáška 03, *data wrangling*): k výsledkům připojíme závod a kvalifikaci.
Cílovou třídu vytvoříme **prahováním** (*thresholding*): `podium = position <= 3`.

```python
df = (results
      .merge(races, on=["season", "round"], how="left")
      .merge(qualifying, on=["season", "round", "driver_id"], how="left"))
df["date"] = pd.to_datetime(df["date"])
df["podium"] = (df["position"] <= 3).astype(int)
df["finished"] = df["status"].str.contains("Finished|Lap", regex=True).astype(int)
print(df.shape, "| duplicity:", df.duplicated(["season", "round", "driver_id"]).sum())
df.head()
```

## 2. Typy atributů
| typ (přednáška 03) | atributy |
|---|---|
| nominální | `driver_id`, `constructor_id`, `circuit_id`, `country`, `status` |
| binární | `podium` (cíl), `finished` |
| ordinální | `grid`, `quali_position`, `position` |
| kvantitativní – diskrétní | `points`, `laps`, `round` |
| kvantitativní – spojité | `time_millis`, časy Q1–Q3 (po převodu na sekundy) |

```python
def lap_to_sec(s):
    """'1:23.456' -> 83.456"""
    if pd.isna(s) or s == "": return np.nan
    m, sec = s.split(":") if ":" in s else ("0", s)
    return int(m) * 60 + float(sec)

for q in ["q1", "q2", "q3"]:
    df[f"{q}_sec"] = df[q].map(lap_to_sec)
df[["season", "round", "grid", "quali_position", "points", "laps", "q1_sec", "podium"]].dtypes
```

## 3. Cílová proměnná – nevyváženost tříd

```python
ax = df["podium"].value_counts().rename({0: "mimo pódium", 1: "pódium"}).plot.bar(color=["#9aa5b1", "#e10600"])
ax.set_title("Rozložení cílové třídy"); ax.set_ylabel("počet instancí"); plt.xticks(rotation=0); plt.show()
print(df["podium"].value_counts(normalize=True).round(3))
```

Pozitivní třída tvoří jen kolem 15 % instancí, třídy jsou tedy **nevyvážené**. Samotná accuracy proto zavádí: model, který vždycky tipne „ne“, má ~85 %. V modelování budeme sledovat precision, recall, F1 a ROC AUC.

## 4. Statistiky polohy, rozptylu a tvaru

```python
num = ["grid", "quali_position", "points", "laps", "q1_sec"]
summary = df[num].describe().T
summary["median"] = df[num].median()
summary["mode"] = df[num].mode().iloc[0]
summary["IQR"] = summary["75%"] - summary["25%"]
summary["skewness"] = df[num].skew()
summary["kurtosis"] = df[num].apply(lambda s: stats.kurtosis(s.dropna(), fisher=False))  # normální rozdělení = 3
summary.round(2)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(df["grid"], bins=21, ax=axes[0]).set_title("Startovní pozice (grid)")
sns.histplot(df.loc[df["points"] > 0, "points"], bins=20, ax=axes[1]).set_title("Body (jen bodující) – pravostranně zešikmené")
sns.histplot(df["q1_sec"].dropna(), bins=40, ax=axes[2]).set_title("Čas Q1 [s] – závisí na trati")
plt.tight_layout(); plt.show()
```

Časy z kvalifikace nejdou porovnávat mezi tratěmi (Monako ~70 s, Spa ~105 s). Je to **kontextová** odlišnost, takže je v preprocessingu převedeme na relativní hodnotu: **ztráta na pole position v %** v rámci závodu.

```python
df["gap_to_pole_pct"] = (df["q1_sec"] / df.groupby(["season", "round"])["q1_sec"].transform("min") - 1) * 100
sns.histplot(df["gap_to_pole_pct"].clip(upper=10), bins=40).set_title("Ztráta na nejrychlejší čas Q1 v závodě [%]")
plt.show()
```

## 5. Vztah atributů k cíli – boxploty a scatter ploty

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.boxplot(data=df[df["grid"] > 0], x="podium", y="grid", ax=axes[0]).set_title("Grid vs. pódium")
p_by_grid = df[df["grid"] > 0].groupby("grid")["podium"].mean()
axes[1].bar(p_by_grid.index, p_by_grid.values, color="#e10600")
axes[1].set_title("Pravděpodobnost pódia podle startovní pozice"); axes[1].set_xlabel("grid")
plt.tight_layout(); plt.show()
```

```python
sns.scatterplot(data=df[df["grid"] > 0].sample(min(3000, len(df)), random_state=1),
                x="grid", y="position", hue="podium", alpha=.4, palette={0: "#9aa5b1", 1: "#e10600"})
plt.title("Startovní vs. cílová pozice"); plt.show()
```

```python
top = df.groupby("constructor_id")["podium"].agg(["mean", "count"]).query("count >= 40").sort_values("mean", ascending=False)
top["mean"].head(12).plot.barh(color="#e10600").invert_yaxis()
plt.title("Podíl pódií podle týmu"); plt.xlabel("P(pódium)"); plt.show()
```

## 6. Korelace a multikolinearita
Pro ordinální atributy použijeme **Spearmanovu** korelaci. `grid` a `quali_position` budou silně korelované. To vadí hlavně lineárním modelům (logistická regrese), takže jeden z nich vyřadíme nebo je sloučíme.

```python
corr = df[["grid", "quali_position", "gap_to_pole_pct", "points", "laps", "round", "podium"]].corr(method="spearman")
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1)
plt.title("Spearmanova korelace"); plt.show()
```

## 7. Chybějící hodnoty

```python
miss = df[["grid", "quali_position", "q1", "q2", "q3", "time_millis"]].isna().mean().mul(100).round(1)
miss.to_frame("% chybí")
```

Chybějící hodnoty tu **nesou informaci**, nejde o náhodu (přednáška 03: *missing value is correlated to the target*):
- `q2`/`q3` chybí jezdcům, kteří vypadli v Q1/Q2. Místo imputace přidáme binární příznak `reached_q3`.
- `time_millis` chybí všem, kdo nedojeli nebo byli dojeti o kolo. Je to známé až po závodě, takže do modelu **nepatří** (únik dat).
- Chybějící kvalifikace (start z boxů, penalizace) → imputace podle `grid`.

```python
df["reached_q3"] = df["q3"].notna().astype(int)
pd.crosstab(df["reached_q3"], df["podium"], normalize="index").round(3)
```

## 8. Odlehlé hodnoty
`grid = 0` znamená start z boxové uličky. Hodnota je platná, ale v ordinální škále nedává smysl (0 není „lepší než 1.“). Převedeme ji na poslední místo a přidáme příznak `pit_lane_start`. Pro `gap_to_pole_pct` použijeme pravidlo **1,5 × IQR** z boxplotu.

```python
print("start z boxů:", (df["grid"] == 0).sum())
q1_, q3_ = df["gap_to_pole_pct"].quantile([.25, .75])
upper = q3_ + 1.5 * (q3_ - q1_)
print(f"horní mez IQR = {upper:.2f} %, odlehlých: {(df['gap_to_pole_pct'] > upper).sum()}")
sns.boxplot(x=df["gap_to_pole_pct"]); plt.title("Ztráta na pole [%] – odlehlé hodnoty"); plt.show()
```

```python
df.to_csv(f"{DATA}/merged.csv", index=False)
print("Uloženo:", df.shape)
```

## Shrnutí pro dokumentaci
- třídy jsou nevyvážené (~15 % pódií) → metriky F1/ROC AUC a `class_weight`
- nejsilnější prediktory: startovní pozice, kvalifikace a tým
- chybějící hodnoty nesou informaci → indikátory (`reached_q3`), žádná slepá imputace
- časy kvalifikace jsou kontextové → relativní `gap_to_pole_pct`
- **pozor na únik dat:** `position`, `points`, `laps`, `status`, `time_millis` jsou známé až po závodě
