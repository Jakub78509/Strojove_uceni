# Modelování – predikce pódia ve F1

Fáze **Modeling** a **Evaluation** procesu CRISP-DM (přednáška 01). Vstupem je tabulka `data/features.csv` z notebooku `Analyza_a_priprava.ipynb`.

**Postup podle přednášek a cvičení:**
1. Rozdělení dat podle času (trénink 2014–2023, test 2024–2026)
2. Baseline modely: „vždy bez pódia“ a „startovní pozice ≤ 3“
3. `Pipeline` s imputací a škálováním až po rozdělení dat – cvičení 4 a 5
4. Logistická regrese – cvičení 5, motivační příklad
5. Rozhodovací strom: kritéria Gini a entropie, prořezávání, ladění hyperparametrů – přednáška 05, cvičení 5
6. Vyhodnocení: matice záměn, precision, recall, F1, ROC křivka – přednáška 04
7. Rozhodovací pravidlo „3 nejpravděpodobnější v závodě“
8. Výsledky po sezónách (vliv nových pravidel 2026) a uložení modelů

Další klasifikátory (naivní Bayes, k-NN, SVM, ensembly) doplníme, až je proberou přednášky.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import joblib

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import GroupKFold, GridSearchCV, cross_validate
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
                             classification_report, confusion_matrix, ConfusionMatrixDisplay, RocCurveDisplay)

sns.set_theme(style="whitegrid")
pd.set_option("display.width", 200)
RANDOM_STATE = 42
DATA = Path("../data")
MODELS = Path("../models"); MODELS.mkdir(exist_ok=True)
```

## 1. Data a rozdělení podle času

Data **nedělíme náhodně** (`train_test_split`), ale podle sezón. Model se učí na minulosti a testuje se na budoucnosti, tedy stejně, jak by se používal v praxi. Náhodné dělení by mohlo dát do tréninku závod z roku 2025 a do testu závod z roku 2024. Model by tak „viděl budoucnost“ a výsledky by byly příliš optimistické.

| množina | sezóny | k čemu slouží |
|---|---|---|
| trénovací | 2014–2023 | učení a ladění (křížová validace) |
| testovací | 2024–2026 | jediné závěrečné vyhodnocení |

```python
df = pd.read_csv(DATA / "features.csv", parse_dates=["date"])

FEATURES = ["grid_clean", "pit_lane_start", "quali_position", "gap_to_pole_pct", "gap_outlier", "reached_q3",
            "driver_form_pts", "driver_podium_rate", "champ_points_before", "champ_rank_before",
            "team_form_pts", "team_rank_before", "team_tech_dnf_rate",
            "driver_starts", "is_rookie", "driver_age", "track_top3_conversion"]
TARGET = "podium"

train = df[df["season"] <= 2023].copy()
test = df[df["season"] >= 2024].copy()
X_train, y_train, g_train = train[FEATURES], train[TARGET], train["season"]
X_test, y_test = test[FEATURES], test[TARGET]

pd.DataFrame({
    "sezóny": ["2014–2023", "2024–2026"],
    "instancí": [len(train), len(test)],
    "závodů": [train.groupby(["season", "round"]).ngroups, test.groupby(["season", "round"]).ngroups],
    "podíl pódií": [f"{y_train.mean():.1%}", f"{y_test.mean():.1%}"],
}, index=["trénovací", "testovací"])
```

Do modelu jdou jen číselné atributy. Kategorie `team`, `circuit_id` a `driver_id` záměrně **nepoužíváme**: model by se učil jména („Red Bull = pódium“) místo obecných vzorců. Jména se navíc mění a noví jezdci a týmy by pro model byli neznámí. Jezdce a tým proto popisují čísla (forma, síla, spolehlivost).

### Jednotná funkce pro vyhodnocení
Podle přednášky 04 sledujeme kromě accuracy hlavně metriky pro **pozitivní třídu** (pódium), protože třídy jsou nevyvážené.

```python
results = []

def evaluate(name, y_true, y_pred, y_proba=None, store=True):
    row = {
        "model": name,
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall": recall_score(y_true, y_pred),
        "F1": f1_score(y_true, y_pred),
        "ROC AUC": roc_auc_score(y_true, y_proba) if y_proba is not None else np.nan,
    }
    if store:
        results.append(row)
    return pd.Series(row).drop("model").astype(float).round(3)
```

## 2. Baseline modely
Nejjednodušší možné modely bez učení. Každý další model musí být lepší, jinak nemá smysl.

```python
evaluate("baseline: vždy bez pódia", y_test, np.zeros(len(y_test), dtype=int))
evaluate("baseline: startovní pozice ≤ 3", y_test, (X_test["grid_clean"] <= 3).astype(int),
         y_proba=-X_test["grid_clean"])   # nižší pozice = vyšší „skóre“ pro ROC
pd.DataFrame(results).set_index("model").round(3)
```

Model „vždy bez pódia“ má accuracy ~85 %, a přitom nepozná ani jedno pódium (recall = 0). Přesně proto podle přednášky 04 accuracy u nevyvážených tříd nestačí. Pravidlo podle startovní pozice je silný baseline s F1 ≈ 0,69.

## 3. Křížová validace podle sezón
Hyperparametry ladíme jen na trénovacích datech pomocí křížové validace (cvičení 5). Místo `KFold` s náhodným zamícháním používáme **`GroupKFold`, kde skupinou je sezóna**. Jeden fold = celé sezóny, takže se závody téže sezóny nikdy nerozdělí mezi trénink a validaci.

```python
cv = GroupKFold(n_splits=5)
for i, (tr, va) in enumerate(cv.split(X_train, y_train, groups=g_train), 1):
    print(f"fold {i}: validační sezóny {sorted(g_train.iloc[va].unique())}")
```

## 4. Logistická regrese
Logistická regrese je **lineární** model: každému atributu přiřadí váhu a součet převede sigmoidou na pravděpodobnost 0–1 (motivační příklad, cvičení 5).

Celé předzpracování je v `Pipeline`, takže se imputace i škálování naučí (`fit`) jen na trénovacích datech, resp. na trénovací části každého foldu (cvičení 4 a 5):
- **imputace mediánem** pro chybějící hodnoty (přednáška 03),
- **z-score škálování** (`StandardScaler`): u logistické regrese nutné, aby váhy byly srovnatelné a regularizace penalizovala všechny atributy stejně (přednáška 03),
- `class_weight="balanced"`: chyby na vzácné třídě (pódium) váží víc, aby model netipoval jen „bez pódia“.

Kvůli **multikolinearitě** (viz analýza: ρ > 0,9) z každé silně korelované dvojice necháváme jen jeden atribut.

```python
LR_FEATURES = [f for f in FEATURES if f not in ("quali_position", "team_rank_before", "driver_form_pts", "champ_points_before")]

log_reg = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(class_weight="balanced", max_iter=2000, random_state=RANDOM_STATE)),
])

cv_lr = cross_validate(log_reg, X_train[LR_FEATURES], y_train, groups=g_train, cv=cv,
                       scoring=["f1", "roc_auc"], return_train_score=True)
print(f"CV F1:      {cv_lr['test_f1'].mean():.3f} ± {cv_lr['test_f1'].std():.3f}")
print(f"CV ROC AUC: {cv_lr['test_roc_auc'].mean():.3f} ± {cv_lr['test_roc_auc'].std():.3f}")

log_reg.fit(X_train[LR_FEATURES], y_train)
proba_lr = log_reg.predict_proba(X_test[LR_FEATURES])[:, 1]
evaluate("logistická regrese", y_test, log_reg.predict(X_test[LR_FEATURES]), proba_lr)
```

```python
coefs = pd.Series(log_reg.named_steps["model"].coef_[0], index=LR_FEATURES).sort_values()
coefs.plot.barh(color=np.where(coefs > 0, "#e10600", "#9aa5b1"), figsize=(8, 5))
plt.title("Váhy logistické regrese (standardizované atributy)"); plt.xlabel("váha"); plt.axvline(0, color="black", lw=.8)
plt.show()
```

Protože jsou atributy standardizované, jde váhy přímo porovnat. Kladná váha zvyšuje pravděpodobnost pódia, záporná ji snižuje. Startovní pozice má velkou zápornou váhu (čím dál vzadu, tím menší šance), forma a pódia z minulých závodů kladnou.

## 5. Rozhodovací strom
Podle přednášky 05 strom v každém uzlu vybere atribut a hranici, které nejlépe oddělí třídy podle míry nečistoty (**Gini** nebo **entropie / informační zisk**). Strom nepotřebuje škálování, potřebuje ale doplnit chybějící hodnoty.

### 5.1 Nepořezaný strom → přeučení

```python
tree_full = Pipeline([("imputer", SimpleImputer(strategy="median")),
                      ("model", DecisionTreeClassifier(random_state=RANDOM_STATE))])
tree_full.fit(X_train, y_train)
t = tree_full.named_steps["model"]
print(f"hloubka: {t.get_depth()}, listů: {t.get_n_leaves()}")
print(f"F1 trénovací: {f1_score(y_train, tree_full.predict(X_train)):.3f}")
print(f"F1 testovací: {f1_score(y_test, tree_full.predict(X_test)):.3f}")
```

Strom bez omezení naroste do velké hloubky a trénovací data se naučí nazpaměť (F1 = 1,0). Na testovacích datech má ale F1 jen kolem 0,5, tedy výrazně méně než jednoduché pravidlo podle startovní pozice. To je **přeučení** (*overfitting*), o kterém mluví přednáška 05: stromy mají vysoký rozptyl a je potřeba je **prořezat**.

### 5.2 Ladění hyperparametrů (pre-pruning a post-pruning)
Grid search s křížovou validací podle sezón (cvičení 5 a 9) vyzkouší:
- `criterion`: Gini vs. entropie (přednáška 05),
- `max_depth`, `min_samples_leaf`: **pre-pruning**, tedy zastavení růstu stromu,
- `ccp_alpha`: **post-pruning** (cost-complexity), tedy odstranění podstromů, které málo zlepšují chybu,
- `class_weight`: vyvážení tříd.

```python
tree_pipe = Pipeline([("imputer", SimpleImputer(strategy="median")),
                      ("model", DecisionTreeClassifier(random_state=RANDOM_STATE))])
param_grid = {
    "model__criterion": ["gini", "entropy"],
    "model__max_depth": [3, 4, 5, 6, 8, None],
    "model__min_samples_leaf": [1, 10, 30, 60],
    "model__ccp_alpha": [0.0, 0.0005, 0.001, 0.003],
    "model__class_weight": [None, "balanced"],
}
grid_tree = GridSearchCV(tree_pipe, param_grid, scoring="f1", cv=cv, n_jobs=-1, return_train_score=True)
grid_tree.fit(X_train, y_train, groups=g_train)

print("nejlepší parametry:", {k.replace("model__", ""): v for k, v in grid_tree.best_params_.items()})
print(f"CV F1: {grid_tree.best_score_:.3f}")
```

```python
cvres = pd.DataFrame(grid_tree.cv_results_)
cvres["max_depth"] = cvres["param_model__max_depth"].map(lambda v: "None" if v is None or pd.isna(v) else str(int(v)))
best_by_depth = (cvres.groupby(["param_model__criterion", "max_depth"])[["mean_train_score", "mean_test_score"]]
                 .max().reset_index())
order = ["3", "4", "5", "6", "8", "None"]
fig, ax = plt.subplots(figsize=(9, 4.5))
for crit, style in [("gini", "-"), ("entropy", "--")]:
    d = best_by_depth[best_by_depth["param_model__criterion"] == crit].set_index("max_depth").loc[order]
    ax.plot(order, d["mean_train_score"], style, marker="o", color="#9aa5b1", label=f"trénovací ({crit})")
    ax.plot(order, d["mean_test_score"], style, marker="o", color="#e10600", label=f"validační ({crit})")
ax.set(title="F1 podle maximální hloubky stromu (nejlepší kombinace ostatních parametrů)", xlabel="max_depth", ylabel="F1")
ax.legend(); plt.show()
```

S rostoucí hloubkou trénovací F1 roste, ale validační od určité hloubky stagnuje nebo klesá. To je typický obraz přeučení: mělčí strom generalizuje lépe. Rozdíl mezi Gini a entropií je malý, což odpovídá přednášce 05 (obě míry nečistoty se chovají podobně).

```python
best_tree = grid_tree.best_estimator_
proba_tree = best_tree.predict_proba(X_test)[:, 1]
evaluate("rozhodovací strom (vyladěný)", y_test, best_tree.predict(X_test), proba_tree)
```

### 5.3 Interpretace stromu
Rozhodovací strom je **white-box** model (přednáška 01). Pro čitelnost vykreslíme jeho horní 3 úrovně a důležitost atributů (o kolik daný atribut celkem snížil nečistotu).

```python
plt.figure(figsize=(22, 9))
plot_tree(best_tree.named_steps["model"], feature_names=FEATURES, class_names=["bez pódia", "pódium"],
          filled=True, rounded=True, max_depth=3, fontsize=9, proportion=True, impurity=False)
plt.title("Vyladěný rozhodovací strom – horní 3 úrovně"); plt.show()
```

```python
imp = pd.Series(best_tree.named_steps["model"].feature_importances_, index=FEATURES).sort_values()
imp[imp > 0].plot.barh(color="#e10600", figsize=(8, 5))
plt.title("Důležitost atributů v rozhodovacím stromu"); plt.xlabel("podíl na snížení nečistoty"); plt.show()
```

## 6. Vyhodnocení na testovacích datech (2024–2026)

```python
summary = pd.DataFrame(results).set_index("model")
summary.style.format("{:.3f}").highlight_max(axis=0, color="#ffd6d6")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))
preds = {"baseline: startovní pozice ≤ 3": (X_test["grid_clean"] <= 3).astype(int),
         "logistická regrese": log_reg.predict(X_test[LR_FEATURES]),
         "rozhodovací strom": best_tree.predict(X_test)}
for ax, (name, p) in zip(axes, preds.items()):
    ConfusionMatrixDisplay(confusion_matrix(y_test, p), display_labels=["bez pódia", "pódium"]).plot(ax=ax, colorbar=False, cmap="Reds")
    ax.set_title(name)
plt.tight_layout(); plt.show()
```

**Jak číst matici záměn** (přednáška 04): řádky jsou skutečnost, sloupce predikce. Vpravo dole jsou správně trefená pódia (TP), vpravo nahoře „falešná pódia“ (FP, snižují precision), vlevo dole přehlédnutá pódia (FN, snižují recall).

Logistická regrese s `class_weight="balanced"` tipuje pódium častěji: má vysoký recall, ale nižší precision. To je volba **prahu**, kterou ukazuje ROC křivka.

```python
fig, ax = plt.subplots(figsize=(6.5, 6))
RocCurveDisplay.from_predictions(y_test, -X_test["grid_clean"], name="baseline: startovní pozice", ax=ax, curve_kwargs={"color": "#9aa5b1"})
RocCurveDisplay.from_predictions(y_test, proba_lr, name="logistická regrese", ax=ax, curve_kwargs={"color": "#1f77b4"})
RocCurveDisplay.from_predictions(y_test, proba_tree, name="rozhodovací strom", ax=ax, curve_kwargs={"color": "#e10600"})
ax.plot([0, 1], [0, 1], "k--", lw=.8, label="náhodné hádání")
ax.set_title("ROC křivky na testovacích datech"); ax.legend(loc="lower right"); plt.show()
```

ROC křivka (přednáška 04) ukazuje model přes **všechny prahy**: čím blíž levému hornímu rohu, tím lépe model odděluje pódia od zbytku. Plocha pod křivkou (ROC AUC) nezávisí na zvoleném prahu.

## 7. Rozhodovací pravidlo „3 nejpravděpodobnější v závodě“
Model dává pravděpodobnost pro každého jezdce zvlášť. S prahem 0,5 se může stát, že v jednom závodě tipne 5 pódií a v jiném žádné. Víme ale, že **v každém závodě jsou na pódiu přesně 3 jezdci**. Využijeme tuto doménovou znalost: v každém závodě označíme za pódium 3 jezdce s nejvyšší pravděpodobností.

```python
def top3_per_race(frame, proba):
    # Shody v pravděpodobnosti (u stromu časté – stejný list) rozhodne startovní pozice, tedy informace známá
    # před závodem. Pořadí řádků použít nesmíme: data jsou seřazená podle cílové pozice → únik dat!
    score = pd.Series(proba, index=frame.index) - frame["grid_clean"] * 1e-6
    rank = score.groupby([frame["season"], frame["round"]]).rank(ascending=False, method="min")
    return (rank <= 3).astype(int).values

evaluate("logistická regrese + top 3 v závodě", y_test, top3_per_race(test, proba_lr), proba_lr)
evaluate("rozhodovací strom + top 3 v závodě", y_test, top3_per_race(test, proba_tree), proba_tree)

summary = pd.DataFrame(results).set_index("model")
summary.style.format("{:.3f}").highlight_max(axis=0, color="#ffd6d6")
```

Pravidlo „top 3“ tipuje vždy přesně tolik pódií, kolik jich skutečně je, takže precision a recall vycházejí stejně. Je to férové srovnání s baselinem, který také tipuje vždy tři jezdce (první trojici na startu).

**Výsledek je důležitý a do dokumentace ho napíšeme poctivě: ani jeden model zatím baseline podle startovní pozice v F1 nepřekonal.** Modely mají o něco vyšší ROC AUC (lépe seřadí jezdce podle šance), ale když z nich uděláme tip „kdo bude na pódiu“, vyjde to podobně nebo hůř než první trojice na startu. Podíváme se, v čem se modely od startovní pozice liší.

```python
def disagreements(model_pick, name):
    d = test.assign(model=model_pick, base=(test["grid_clean"] <= 3).astype(int))
    only_model = d[(d["model"] == 1) & (d["base"] == 0)]   # model tipne jezdce mimo první trojici
    only_base = d[(d["model"] == 0) & (d["base"] == 1)]    # model vyřadí jezdce z první trojice
    return pd.Series({
        "případů, kdy se liší": len(only_model),
        "tip modelu správně": f"{only_model['podium'].mean():.0%}",
        "tip podle startu správně": f"{only_base['podium'].mean():.0%}",
    }, name=name)

pd.concat([disagreements(top3_per_race(test, proba_lr), "logistická regrese"),
           disagreements(top3_per_race(test, proba_tree), "rozhodovací strom")], axis=1)
```

Z 195 tipů (65 závodů × 3) se logistická regrese od startovní pozice liší ve 47 případech a strom ve 20. Typicky „povýší“ silného jezdce ze 4.–5. místa (dobrá forma, silný tým) na úkor jezdce z první trojice na startu. Tyto záměny ale vycházejí **hůř** než startovní pozice: jezdec tipnutý modelem dojel na pódiu zhruba ve třetině případů, vyřazený jezdec z první trojice zhruba v polovině. Startovní pozice už v sobě obsahuje většinu informace o rychlosti auta i jezdce, protože kvalifikace se jede den před závodem. Forma a síla týmu jsou starší informace a modely jim přikládají příliš velkou váhu.

## 8. Výsledky po sezónách
Sezóna 2026 přinesla nová technická pravidla. Podíváme se, jestli se kvalita modelů v testovacích sezónách liší.

```python
test = test.assign(base=(test["grid_clean"] <= 3).astype(int),
                   lr_top3=top3_per_race(test, proba_lr),
                   tree_top3=top3_per_race(test, proba_tree))
by_season = test.groupby("season").apply(lambda d: pd.Series({
    "závodů": d.groupby("round").ngroups,
    "baseline (grid ≤ 3)": f1_score(d["podium"], d["base"]),
    "log. regrese top 3": f1_score(d["podium"], d["lr_top3"]),
    "strom top 3": f1_score(d["podium"], d["tree_top3"]),
}), include_groups=False)
by_season.round(3)
```

## 9. Uložení modelů
Modely uložíme s datem, ke kterému jsou zafixovaná data (17. 10. 2026). Na nich pak uděláme predikce zbývajících závodů sezóny 2026 a porovnáme je se skutečnými výsledky.

```python
joblib.dump({"model": log_reg, "features": LR_FEATURES}, MODELS / "logisticka_regrese_2026-10-17.joblib")
joblib.dump({"model": best_tree, "features": FEATURES}, MODELS / "rozhodovaci_strom_2026-10-17.joblib")
sorted(p.name for p in MODELS.glob("*.joblib"))
```

| rozhodovací strom (vyladěný, práh 0,5) | 0,668 | 0,939 |

- **Accuracy u nevyvážených tříd zavádí** (přednáška 04): model „vždy bez pódia“ má 85 % a nepozná nic.
- **Logistická regrese** s vyvážením tříd má vysoký recall (93 %), ale nízkou precision (46 %). Váhy ukazují, že kromě startovní pozice přispívá forma jezdce a síla týmu.
- **Nepořezaný strom se přeučí** (trénovací F1 = 1,0, testovací 0,53). Po prořezání se zlepší na 0,67 a zůstává čitelný (přednáška 05). Grid search vybral entropii, hloubku 5 a minimálně 30 instancí v listu.
- **Modely zatím nepřekonaly baseline v F1**, i když mírně lépe řadí jezdce podle šance (vyšší ROC AUC). Startovní pozice už obsahuje většinu informace, kterou forma a síla týmu nesou. Je to legitimní a zajímavé zjištění, které v dokumentaci rozebereme.
